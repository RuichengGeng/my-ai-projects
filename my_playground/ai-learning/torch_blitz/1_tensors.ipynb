```python
import torch.nn as nn
import torch.nn.functional as F
```

```python
class MyNet(nn.Module):
    def __init__(self):
        super().__init__()
        self.fc1 = nn.Linear(1, 5)
    
    def forward(self, x):
        C1 = self.fc1(x)
        C2 = C1 * C1
        return C2
```

```python

net = MyNet()
print(net)
```

```python
out = net.forward(torch.tensor([10.0]))
target = torch.rand(5)
target = target * target * 5
```

```python
out.view(1, -1)
```

```python
target.view(1, -1)
```

```python
criterion = nn.MSELoss()
loss = criterion(out, target)
print(loss)
```

```python
import torch.optim as optim

# create your optimizer
optimizer = optim.SGD(net.parameters(), lr=0.01)

```

```python
net.parameters()
```

```python
for i in range(10):                                                                                                                                                       
    optimizer.zero_grad()                                                                                                                                                   
    out = net(torch.tensor([10.0]))     # recompute every step                                                                                                              
    loss = criterion(out, target)                                                                                                                                           
    loss.backward()                                                                                                                                                         
    optimizer.step()                                                                                                                                                        
    print(loss.item())   
```
